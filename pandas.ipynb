```python
data_with_missing = {
  'cars': ["BMW", "Volvo", None, "Toyota", "Merc", "Kia"],
  'ratings': [8, 7, 6, None, 7, 6]
}
df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```
